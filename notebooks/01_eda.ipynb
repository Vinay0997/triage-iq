# TriageIQ - 01 EDA
Exploratory data analysis of the synthetic 311 dataset.

Expected location: `triage-iq/notebooks/`, data at `../data/311_requests.csv`.

## 1. Load data

```
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
%matplotlib inline
df = pd.read_csv("../data/311_requests.csv", parse_dates=["created_date"])
print(df.shape)
df.head()
```

## 2. Structure and missing values

```
df.info()
missing = df.isna().mean().sort_values(ascending=False)
missing[missing > 0]
```

## 3. Target balance (category and request_type)

```
print(df["category"].value_counts(normalize=True).round(3))
df["request_type"].value_counts().plot(kind="barh", figsize=(8,6), title="Request types")
plt.show()
```

## 4. Time patterns

```
df["hour"] = df.created_date.dt.hour
df["weekday"] = df.created_date.dt.day_name()
df["month"] = df.created_date.dt.to_period("M")
df.groupby("month").size().plot(title="Requests per month", figsize=(10,4))
plt.show()
df.groupby("hour").size().plot(kind="bar", title="Requests by hour", figsize=(10,4))
plt.show()
```

## 5. Resolution time

```
print(df.groupby("category")["resolution_hours"].agg(["count","median","mean"]).round(1))
sns.boxplot(data=df, x="category", y="resolution_hours")
plt.xticks(rotation=30); plt.show()
```

## 6. Channel, priority, status

```
for c in ["channel","priority","status","agency"]:
    print(df[c].value_counts(normalize=True).round(3), "\n")
pd.crosstab(df["category"], df["priority"], normalize="index").round(2)
```

## 7. Text description

```
df["desc_len"] = df["description"].str.len()
print(df["desc_len"].describe())
df.groupby("category")["description"].apply(lambda s: s.sample(3, random_state=1).tolist())
```

## 8. Geography

```
sns.scatterplot(data=df.sample(3000, random_state=1), x="longitude", y="latitude", hue="category", s=10)
plt.show()
df["neighborhood"].value_counts(dropna=False)
```
